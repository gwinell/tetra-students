## 1. Зачем нужны функции

Представь, что программа несколько раз здоровается с пользователем. Если каждый раз писать одни и те же команды, код повторяется. Функция позволяет дать действию имя, записать его один раз и запускать по этому имени.

![Повторяющийся код превращается в функцию](assets/repeated-code-to-function.jpg)

Посмотри на повторяющиеся команды. Пока здесь нет ничего нового: мы уже знаем `print()`.


````python
print("Привет!")
print("Привет!")
print("Привет!")
````

Теперь опишем действие один раз. Строки внутри функции сами по себе не выполняются при создании функции. Они начнут выполняться, когда мы вызовем её по имени.


````python
def say_hello():
    print("Привет!")

say_hello()
say_hello()
say_hello()
````

**Мини-эксперимент.** Измени приветствие на своё и запусти ячейку. Сколько раз появится сообщение? Удали один вызов `say_hello()` и проверь ещё раз.

## 2. Как устроено объявление функции

Ключевое слово `def` сообщает Python: «сейчас я создаю функцию». Затем пишут имя, круглые скобки и двоеточие. Команды функции располагают ниже и с отступом.

![Части объявления функции](assets/function-anatomy.jpg)

Пока запомни форму:

```python
def имя_функции():
    команда
```

Имя обычно описывает действие: `say_hello`, `show_menu`, `print_rules`. В Python принято писать его маленькими английскими буквами, разделяя слова нижним подчёркиванием.


````python
def show_rules():
    print("1. Читай задание внимательно")
    print("2. Пробуй и проверяй")

show_rules()
````

**Важно про отступ.** Команды с одинаковым отступом относятся к телу функции. Если убрать отступ у `print()`, Python не поймёт, что эта команда часть функции.

**Предскажи результат.** Сколько строк выведется, если вызвать `show_rules()` два раза? Проверь, добавив второй вызов.

## 4. Параметры и аргументы

Функция может получать данные, чтобы работать с разными значениями. В объявлении функции имена внутри скобок называются **параметрами**. При вызове мы передаём конкретные значения — это **аргументы**.

Можно представить это как пустое место для значения: параметр `name` принимает аргумент, например строку `"Маша"`.


````python
def greet(name):
    print(f"Привет, {name}!")

greet("Маша")
greet("Тимур")
````

Здесь `name` — параметр, а `"Маша"` и `"Тимур"` — аргументы. При каждом вызове функция получает своё значение.

**Мини-эксперимент.** Вызови `greet()` ещё раз со своим именем. Что будет, если в вызове забыть написать значение? Убедись, что для функции с параметром нужен аргумент.

### Несколько параметров

В скобках можно указать несколько параметров, разделяя их запятыми. Аргументы передаются в том же порядке: первый — первому параметру, второй — второму.


````python
def show_sum(first_number, second_number):
    print(first_number + second_number)

show_sum(7, 5)
show_sum(20, 3)
````

В первом вызове `7` попадает в `first_number`, а `5` — в `second_number`.

## 5. Локальные переменные — имена внутри функции

Переменные, которые создаются внутри функции, помогают ей выполнять работу. Например, переменная `total` ниже нужна для вычисления суммы. Пока достаточно помнить: она создаётся при выполнении функции и используется в её командах.


````python
def show_double(number):
    doubled_number = number * 2
    print("Удвоенное число:", doubled_number)

show_double(6)
show_double(15)
````

## 6. `return`: функция отдаёт результат

До сих пор функции **показывали** ответ командой `print()`. Но часто нужно не просто увидеть ответ на экране, а продолжить использовать его в программе.

`return` возвращает значение из функции. После выполнения `return` функция заканчивает свою работу.

![Аргументы, функция, return и сравнение с print](assets/arguments-return-print.jpg)

Сначала сравним две функции. Первая только печатает сумму. Вторая возвращает сумму, чтобы программа могла взять её дальше.


````python
def show_sum(first, second):
    print(first + second)

def calculate_sum(first, second):
    return first + second

show_sum(4, 6)
result = calculate_sum(4, 6)
print("Сохранённый результат:", result)
````

Вызов `show_sum(4, 6)` показывает `10`, но не отдаёт это число для присваивания переменной. А `calculate_sum(4, 6)` возвращает число `10`, и мы сохраняем его в `result`.

`print()` отвечает на вопрос **«что показать на экране?»**. `return` отвечает на вопрос **«какое значение функция отдаёт программе?»**.

### Используем возвращённое значение

Вернувшееся значение можно сохранить в переменную, вставить в f-строку, сравнить в `if` или передать другой функции.


````python
def calculate_sum(first, second):
    return first + second

answer = calculate_sum(8, 5)
print("Сумма равна", answer)
print(f"Если сложить 8 и 5, получится {answer}")

if answer > 10:
    print("Сумма больше 10")
else:
    print("Сумма не больше 10")
````

Значение функции можно передать сразу в другую функцию. Сначала Python вычислит `calculate_sum(2, 3)`, затем передаст результат в `print()`.


````python
def calculate_sum(first, second):
    return first + second

print(calculate_sum(2, 3))
````

**Проверь себя.** Что произойдёт, если написать `return first + second` вместо `print(first + second)`? А если вызвать функцию без `print()` и без сохранения результата?

**Попробуй.** Напиши `multiply(first, second)`, которая возвращает произведение. Сохрани результат в переменную и выведи сообщение с f-строкой.

### Частая ошибка: `print()` внутри функции вместо `return`

Если функция печатает значение, присваивание получает не напечатанное число. Оно получает специальное значение `None`, потому что функция ничего не вернула.


````python
def show_double(number):
    print(number * 2)

printed_value = show_double(5)
print("В переменной:", printed_value)
````

Теперь заменим `print()` внутри функции на `return`. Результат можно сохранить и использовать дальше.


````python
def get_double(number):
    return number * 2

returned_value = get_double(5)
print("В переменной:", returned_value)
print("Ещё раз удвоим:", returned_value * 2)
````

Запомни коротко:

- `print(значение)` показывает значение на экране;
- `return значение` отдаёт значение туда, где функцию вызвали;
- если функция дошла до конца и не встретила `return`, она не возвращает полезное значение.

**Мини-эксперимент.** В `get_double()` верни не удвоенное, а утроенное число. Что напечатает последняя строка?

## 7. Функции с условиями, циклами и списками

Функция — это способ дать имя набору команд. Внутри неё работают знакомые нам `if`, `for`, `while` и списки.

### Условие внутри функции


````python
def describe_age(age):
    if age < 13:
        return "ребёнок"
    else:
        return "подросток или взрослый"

print(describe_age(10))
print(describe_age(14))
````

**Попробуй.** Добавь ветку `elif`, чтобы отдельно описать возраст 13 лет. Выбери порядок условий так, чтобы каждое значение попадало в нужную ветку.

### Цикл `for` и список внутри функции

Передадим функции список оценок. Внутри она проверит каждую оценку и посчитает, сколько из них не меньше четырёх.


````python
def count_good_scores(scores):
    count = 0
    for score in scores:
        if score >= 4:
            count = count + 1
    return count

lesson_scores = [5, 3, 4, 2, 5]
good_count = count_good_scores(lesson_scores)
print("Оценок 4 или 5:", good_count)
````

Параметр `scores` получает список, который передан в вызове. Переменная `count` начинает со значения `0`, цикл увеличивает её, а `return` отдаёт готовое количество.

**Мини-задание.** Измени список и проверь подсчёт. Измени условие так, чтобы считать оценки меньше четырёх.

### Создаём новый список

Функция может вернуть новый список. Здесь мы выберем из исходного списка только положительные числа.


````python
def get_positive_numbers(numbers):
    positive_numbers = []
    for number in numbers:
        if number > 0:
            positive_numbers.append(number)
    return positive_numbers

values = [-3, 8, 0, 4, -1, 6]
result = get_positive_numbers(values)
print(result)
````

**Попробуй.** Измени условие на `number % 2 == 0`, чтобы функция собирала чётные числа. Для нуля это условие тоже верно.

### `while` внутри функции

Функция также может повторять команды с помощью `while`. В этом примере она складывает числа от `1` до `limit`.


````python
def sum_to(limit):
    number = 1
    total = 0
    while number <= limit:
        total = total + number
        number = number + 1
    return total

print(sum_to(5))
print(sum_to(10))
````

## 8. Модуль — отдельный файл с функциями

Когда функций становится много, их удобно хранить отдельно от основной программы. Файл Python с командами и функциями называется **модулем**. Например, мы создадим `student_tools.py`, а затем будем использовать его из ноутбука.

В Google Colab следующая ячейка создаст настоящий файл рядом с ноутбуком в текущей папке. В репозитории тот же файл тоже лежит рядом с этим блокнотом.

![Ноутбук и модуль в одной папке](assets/module-import-flow.jpg)

Запусти эту ячейку, чтобы создать модуль в Colab или в текущем окружении Jupyter.


````python
%%writefile student_tools.py

def greet(name):
    """Возвращает приветствие для указанного имени."""
    return f"Привет, {name}!"


def add_numbers(first, second):
    """Возвращает сумму двух чисел."""
    return first + second


def is_adult(age):
    """Возвращает True, если возраст не меньше 18."""
    if age >= 18:
        return True
    else:
        return False


def count_long_words(words, minimum_length):
    """Считает слова, длина которых не меньше заданной."""
    count = 0
    for word in words:
        if len(word) >= minimum_length:
            count = count + 1
    return count
````

## 9. Подключаем модуль командой `import`

Команда `import student_tools` подключает файл `student_tools.py`. После неё к функции обращаются через имя модуля, точку и имя функции: `student_tools.greet(...)`.

Важно для простого учебного сценария: `student_tools.py` должен быть доступен Python в той же папке или текущем рабочем каталоге. Запускай ячейки сверху вниз, чтобы сперва создать файл.


````python
import student_tools

print(student_tools.greet("Маша"))
print(student_tools.add_numbers(7, 5))
print(student_tools.is_adult(16))
````

Точка в `student_tools.greet` означает: «возьми функцию `greet` из модуля `student_tools`».

**Мини-эксперимент.** Замени имя и числа. Попробуй вызвать `student_tools.count_long_words()` со списком слов и числом минимальной длины.


````python
words = ["кот", "черепаха", "ёж", "тигр"]
long_words_count = student_tools.count_long_words(words, 5)
print("Длинных слов:", long_words_count)
````

### Импортируем выбранную функцию

Если нужна конкретная функция, можно написать `from student_tools import greet`. Тогда имя функции доступно напрямую, без приставки `student_tools.`.


````python
from student_tools import greet, add_numbers

print(greet("Саша"))
total = add_numbers(12, 8)
print("Сумма:", total)
````

Оба способа используют один и тот же модуль:

```python

```

или

```python

greet("Саша")
```

Пока выбирай тот вариант, который проще читать. Запись `import student_tools as tools` позволяет дать модулю короткое имя, но в этом уроке можно обойтись без неё.

## 10. Маленькая программа с собственным модулем

Теперь объединим знакомые условия, циклы и списки с функциями из `student_tools.py`. Главная программа хранит список учеников, а модуль помогает посчитать слова и поприветствовать ученика.


````python
import student_tools

students = ["Маша", "Тимур", "Алекс"]

for student in students:
    print(student_tools.greet(student))

words = ["Python", "код", "функция", "модуль"]
count = student_tools.count_long_words(words, 5)

if count > 0:
    print(f"Нашлось длинных слов: {count}")
else:
    print("Длинных слов нет")
````
