## Карта урока

В Python строка и список состоят из элементов, которые стоят по порядку. У каждого элемента есть номер — **индекс**. На схемах ниже каждый прямоугольник — это один символ или один элемент списка.

![Индексы строки и списка: положительные и отрицательные](assets/indices.png)

Сначала разберём индексы на строке, затем найдём такие же позиции в списке.

## 1. Строка — это последовательность символов

В строке символы расположены по порядку. У каждого символа есть своя позиция. В Python позицию называют **индексом**.

Важно: счёт индексов начинается с нуля. Поэтому у слова `Python` буква `P` находится на индексе `0`, а не `1`.

```python
word = "Python"
print(word[0])  # P
print(word[1])  # y
print(word[5])  # n
```

### Положительные и отрицательные индексы

Счёт слева направо начинается с `0`. Поэтому буква `P` имеет номер `0`, а `y` — номер `1`. Можно считать и от правого края: последний символ имеет индекс `-1`, перед ним — `-2`.

Для `Python`: `word[0]` и `word[-6]` — один и тот же символ `P`; `word[5]` и `word[-1]` — `n`.

На схеме индексы стоят прямо над и под символами. Сверься с ней перед запуском кода.

![Положительные и отрицательные индексы строки и списка](assets/indices.png)

```python
word = "Python"
print(word[0], word[-6])
print(word[5], word[-1])
print(word[-2])
```

**Мини-эксперимент.** Не запуская, предположи, что напечатает `word[-3]`. Затем измени индекс и проверь.

Если обратиться к несуществующей позиции, Python сообщит об ошибке `IndexError`. Например, у строки из 6 символов нет индекса `6`: последний положительный индекс — `5`.

```python
word = "Python"
print(word[6])  # Попробуй запустить и прочитай сообщение об ошибке
```

## 2. Список — несколько значений под одним именем

Строка хранит символы, а **список** хранит несколько элементов в заданном порядке. Список записывают в квадратных скобках, элементы разделяют запятыми.

Например, список покупок — это одна переменная, в которой лежат несколько названий.

```python
shopping = ["хлеб", "молоко", "яблоки"]
print(shopping)
print(shopping[0])
```

### Индексы работают так же, как в строке

В списке каждый элемент занимает свою позицию. Нумерация такая же, как у букв в строке: первый элемент — индекс `0`; последний — индекс `-1`.

Список `["хлеб", "молоко", "яблоки"]`: `shopping[0]` — хлеб, `shopping[1]` — молоко, `shopping[-1]` — яблоки. Индекс — это номер позиции, а не само значение.

На рисунке та же нумерация показана для строки и списка. Например, первый элемент списка имеет индексы `0` и `-6`, а последний — `5` и `-1`.

![Одинаковая нумерация позиций в строке и списке](assets/indices.png)

```python
shopping = ["хлеб", "молоко", "яблоки"]
print(shopping[0])
print(shopping[-1])
print(shopping[1], shopping[-2])
```

**Попробуй.** Создай список из четырёх любимых игр, книг или блюд. Выведи первый, второй и последний элементы. Для списка из `n` элементов последний положительный индекс равен `n - 1`, а последний отрицательный — всегда `-1`.

```python
favorites = ["шахматы", "робот", "пицца", "космос"]
print(favorites[0])
print(favorites[1])
print(favorites[-1])
```

## 3. `len()` считает количество

Функция `len()` возвращает количество символов в строке или элементов в списке. Она не считает с нуля: в слове `Кот` три символа, поэтому `len("Кот")` равно `3`.

У строки из `n` символов положительные индексы идут от `0` до `n - 1`. Поэтому последнюю позицию часто находят так: `len(строка) - 1`.

```python
word = "Python"
shopping = ["хлеб", "молоко", "яблоки"]
print(len(word))
print(len(shopping))
print(word[len(word) - 1])
```

**Мини-эксперимент.** Добавь элемент в список или поменяй строку. Как изменится результат `len()`? Посчитай заранее, затем запусти ячейку.

```python
word = "Привет!"
colors = ["синий", "зелёный", "жёлтый", "красный"]
print(len(word))
print(len(colors))
```

## 4. Заменяем элемент по индексу

Список можно менять. Чтобы заменить значение в конкретной позиции, слева от `=` укажи список и индекс: `имя_списка[индекс] = новое_значение`. Меняется элемент на этой позиции, остальные остаются на месте.

У строк так сделать нельзя: строки нельзя менять по отдельным индексам. Например, `word[0] = "п"` приведёт к ошибке. Сейчас сосредоточимся на изменяемых списках.

![Замена элемента на позиции с индексом 1](assets/replace-index.png)

Слева показан список до замены, справа — после. Команда `animals[1] = "тигр"` меняет только значение на второй позиции (индекс `1`). Первый и последний элементы остаются на своих местах.

```python
colors = ["синий", "зелёный", "жёлтый"]
colors[1] = "фиолетовый"
print(colors)
```

**Предскажи результат.** Какой список получится? Замени индекс `0` на другой и снова запусти ячейку.

```python
animals = ["кот", "лиса", "заяц"]
animals[0] = "тигр"
print(animals)
```

## 5. Срезы: берём несколько элементов

Индекс даёт один элемент. **Срез** выбирает часть последовательности: строки или списка. Общая запись — `последовательность[start:stop]`.

`start` — индекс, с которого начинаем. `stop` — позиция, на которой останавливаемся, **не включая её**. Это правило «начало включено, конец не включён» помогает точно задавать длину среза. Срез создаёт новую строку или новый список; исходное значение само по себе не удаляется.

![Срез строки: индексы 1, 2 и 3 входят в результат, индекс 4 — нет](assets/slices.png)

В примере `word[1:4]` Python берёт элементы с индексами `1`, `2`, `3`. До позиции `4` доходит, но сам элемент с индексом `4` не берёт.

```python
word = "Python"
print(word[1:4])  # yt h без пробелов: символы с индексами 1, 2, 3

colors = ["синий", "зелёный", "жёлтый", "красный"]
print(colors[1:3])  # элементы с индексами 1 и 2
```

Можно не указывать одну из границ. `[:stop]` берёт с начала до `stop` (не включая его), а `[start:]` — от `start` до конца. Полный срез `[:]` копирует всю последовательность. Отрицательные индексы в границах среза тоже разрешены.

```python
word = "Python"
colors = ["синий", "зелёный", "жёлтый", "красный"]
print(word[:3])
print(word[3:])
print(colors[-2:])
print(colors[:])
```

### Добавляем шаг: `start:stop:step`

Три части среза — это начало, граница остановки и шаг. Шаг показывает, через сколько позиций брать следующий элемент. `step=1` — каждый элемент; `step=2` — через один. Чтобы пройти в обратную сторону, используют шаг `-1`.

Если шаг не задан, Python считает его равным `1`. Шаг не может быть равен нулю.

```python
word = "ПРОГРАММИРОВАНИЕ"
print(word[0:10:2])
print(word[::-1])

numbers = [0, 1, 2, 3, 4, 5, 6]
print(numbers[::2])
print(numbers[::-1])
```

**Мини-задание.** Возьми срез списка `numbers`, чтобы получить `[2, 3, 4]`. Подумай, какой индекс поставить в `start` и какой — в `stop`. Затем возьми элементы через один.

```python
numbers = [0, 1, 2, 3, 4, 5, 6]
print(numbers[2:5])
print(numbers[::2])
```

## 6. Добавляем один элемент: `append()`

`append()` добавляет **один объект** в конец списка. В скобках указывают то, что нужно добавить. Исходный список меняется.

Если добавить строку, добавится вся строка одним элементом, а не отдельные буквы.

![Операции append, insert, extend, remove и del](assets/operations.png)

В первой строке схемы `append()` добавляет один элемент справа, в конец списка.

```python
colors = ["синий", "зелёный"]
colors.append("жёлтый")
print(colors)

letters = ["а", "б"]
letters.append("вг")
print(letters)
```

## 7. Вставляем элемент: `insert()`

`insert()` вставляет один элемент перед указанным индексом. Сначала указывают индекс, затем значение: `список.insert(индекс, значение)`. После вставки элементы справа сдвигаются на одну позицию.

Если индекс равен `0`, элемент станет первым. Чтобы добавить в конец, обычно проще использовать `append()`.

![Вставка элемента методом insert](assets/operations.png)

Во второй строке схемы `insert(1, "лиса")` ставит лису на позицию `1`. Прежний элемент справа сдвигается.

```python
colors = ["синий", "жёлтый"]
colors.insert(1, "зелёный")
print(colors)
```

## 8. `append()` и `extend()` — в чём разница?

`append(значение)` добавляет один объект как один элемент. `extend(последовательность)` проходит по последовательности и добавляет в список каждый её элемент.

Для строк это особенно заметно: `append("кот")` добавит один элемент `"кот"`, а `extend("кот")` добавит три элемента: `"к"`, `"о"`, `"т"`. Обычно `extend()` применяют, чтобы добавить элементы другого списка.

![Чем отличаются append, insert и extend](assets/operations.png)

Схема показывает расположение элементов. Разницу между добавлением одного объекта и добавлением нескольких элементов проверим на коде ниже.

```python
first = ["кот", "пёс"]
first.append(["лиса", "заяц"])
print(first)  # список стал содержать вложенный список как один элемент

second = ["кот", "пёс"]
second.extend(["лиса", "заяц"])
print(second)  # лиса и заяц стали отдельными элементами
```

**Проверь себя.** Что будет внутри `items` после `items.append([3, 4])`? А после `items.extend([3, 4])`? Запусти обе строки с разными копиями списка.

Для урока достаточно запомнить: «один объект» → `append`; «добавить элементы последовательности по одному» → `extend`. `expand()` в списках Python нет: вероятно, имелся в виду именно `extend()`.

```python
a = [1, 2]
b = [1, 2]
a.append([3, 4])
b.extend([3, 4])
print(a)
print(b)
```

## 9. Удаляем значение: `remove()`

`remove(значение)` удаляет **первое совпадение** с указанным значением. Здесь указывают само значение, а не его индекс. Если одинаковое значение встречается несколько раз, за один вызов удалится только первое.

Если такого значения в списке нет, Python сообщит об ошибке `ValueError`. Перед удалением можно проверить список глазами или использовать изученные условия и циклы — это отдельная практика.

![remove удаляет по значению, del — по позиции](assets/operations.png)

На схеме `remove` и `del` показаны отдельно: один ищет значение, другой удаляет элемент на указанной позиции.

```python
colors = ["синий", "зелёный", "жёлтый", "зелёный"]
colors.remove("зелёный")
print(colors)  # удалено только первое совпадение
```

## 10. Удаляем по позиции: `del`

`del` — это команда Python, а не метод списка. После `del` указывают список и индекс: `del список[индекс]`. Так удаляется элемент с нужной позиции, даже если одинаковые значения встречаются ещё где-то.

Можно удалить и часть списка срезом: `del список[start:stop]`. Граница `stop`, как и в обычном срезе, не включается.

![Сравнение remove и del](assets/operations.png)

`remove("зелёный")` ищет значение `"зелёный"`. `del colors[1]` удаляет элемент на позиции `1`.

```python
colors = ["синий", "зелёный", "жёлтый", "зелёный"]
del colors[1]
print(colors)

del colors[1:3]
print(colors)
```

### `remove()` или `del`?

- Знаешь **значение**, которое нужно убрать? Используй `remove(значение)`. Удалится первое совпадение.
- Знаешь **позицию**, которую нужно убрать? Используй `del список[индекс]`.

Оба действия меняют исходный список.

```python
names = ["Аня", "Борис", "Вера"]
names.remove("Борис")  # убрать значение
print(names)

del names[0]  # убрать элемент с индексом 0
print(names)
```

## 11. Находим позицию: `index()`

`index(значение)` возвращает индекс **первого** элемента с таким значением. Метод ничего не удаляет и не меняет.

У строк тоже есть `index()`: он ищет позицию части строки. Но помни: если искомого значения нет, возникнет `ValueError`. `find()` у списков Python нет. Для списка здесь используем `index()`.

```python
colors = ["синий", "зелёный", "жёлтый", "зелёный"]
print(colors.index("жёлтый"))
print(colors.index("зелёный"))  # индекс первого совпадения
```

**Мини-эксперимент.** Измени поиск на значение, которого нет, например `"красный"`, и запусти. Прочитай ошибку. Позже можно будет научиться проверять наличие значения, но сегодня используем `index()` только когда уверены, что элемент есть.

```python
word = "Python"
print(word.index("t"))

colors = ["синий", "зелёный", "жёлтый"]
print(colors.index("зелёный"))
```

## 12. Собираем операции вместе

В следующем примере применены замена по индексу, добавление, вставка, поиск и удаление. После каждой операции печатаем список, чтобы видеть изменения. Читай команды сверху вниз.

```python
tasks = ["прочитать", "погулять", "порисовать"]
print(tasks)

tasks[1] = "погулять с собакой"
print(tasks)

tasks.append("сделать уроки")
print(tasks)

tasks.insert(1, "позавтракать")
print(tasks)

print("Индекс рисования:", tasks.index("порисовать"))
tasks.remove("прочитать")
print(tasks)
del tasks[-1]
print(tasks)
```

## Короткое повторение

- Индексы начинаются с `0`; индекс `-1` обозначает последний элемент.
- `len()` считает символы строки или элементы списка.
- `последовательность[start:stop:step]` выбирает срез; `stop` не входит в результат.
- Списки можно менять по индексу: `items[index] = value`.
- `append()` добавляет один объект; `extend()` добавляет элементы последовательности по одному.
- `insert(index, value)` вставляет значение перед позицией.
- `remove(value)` удаляет первое совпадение; `del items[index]` удаляет позицию.
- `index(value)` находит индекс первого совпадения. У `list` нет `find()` и `expand()`.

**Полезная привычка:** перед обращением по индексу убедись, что такая позиция существует. Если длина списка `len(items)`, последний положительный индекс — `len(items) - 1`.
