В первом уроке мы учились работать с текстом и вводом. Теперь научимся получать от человека число, сравнивать значения и выбирать действия программы.

## 1. Числа и вычисления

Числа в Python записывают без кавычек. С целыми числами можно выполнять привычные действия: `+` сложение, `-` вычитание, `*` умножение. Если заключить цифры в кавычки, получится текст, а не число.


```python
print(8 + 3)
print(8 - 3)
print(8 * 3)

print("8" + "3")  # это соединение текста: получится 83
```

**Мини-эксперимент.** Предскажи результат `4 + 6` и `"4" + "6"`, затем запусти код.

**Задание.** Выведи сумму двух чисел и их произведение. Подставь свои значения.


```python
first_number = 12
second_number = 5
print(first_number + second_number)
print(first_number * second_number)
```

## 2. `input()` возвращает строку

Команда `input()` показывает подсказку и ждёт ответ. Введённое значение сначала становится строкой — даже если человек набрал цифры. Поэтому `input()` не может само по себе сообщить Python, что с ответом нужно считать.


```python
age_text = input("Сколько тебе лет? ")
print("Ты ввёл(а):", age_text)
print("Тип пока текстовый: цифры заключены в кавычки при вводе")
```

**Мини-эксперимент.** Введи `12`. Что будет, если соединить ответы `"1"` и `"2"` как текст?

**Задание.** Попроси пользователя ввести номер любимого числа и выведи ответ с пояснением.


```python
first_digit = input("Введи цифру: ")
second_digit = input("Введи ещё одну цифру: ")
print(first_digit + second_digit)
```

## 3. Преобразуем ввод в целое число: `int()`

`int()` превращает строку, в которой записано целое число, в число. Например, `int("12")` даёт число `12`. После преобразования `+` выполняет сложение.

Введи целое число: допустимы, например, `7`, `0` и `-3`. Слова и дроби вроде `3.5` пока не подходят.

![input → строка → int → число](images/input-int.png)


```python
number_text = input("Введи целое число: ")
number = int(number_text)
print(number + 1)
```

Можно записать короче: поместить `input()` внутрь `int()`. Порядок такой: Python сначала получает ответ, затем преобразует его.


```python
number = int(input("Введи число: "))
print("Через два шага будет:", number + 2)
```

**Мини-эксперимент.** Замени `int` на `str`? Пока не запускать: подумай, почему `str` не проходили. Вместо этого введи в поле `три` и посмотри, какое сообщение появится. Оно означает, что текст нельзя преобразовать в целое число.

**Задание.** Попроси ввести два целых числа. Преобразуй оба и выведи их сумму.


```python
first_number = int(input("Первое число: "))
second_number = int(input("Второе число: "))
print("Сумма:", first_number + second_number)
```

## 4. Сравнения дают ответ True или False

Сравнение проверяет высказывание. Его результат — одно из двух специальных значений: `True` (верно) или `False` (неверно). Это не слова в кавычках.

`==` означает «равно». Один знак `=` сохраняет значение в переменную, два знака `==` сравнивают значения.

![Карточка операторов сравнения](images/comparisons.png)


```python
print(5 == 5)
print(5 != 2)
print(7 > 9)
print(4 <= 4)
```

**Мини-эксперимент.** Замени `==` на `=` внутри `print`. Python сообщит об ошибке: `=` сохраняет значение, но не является сравнением.

**Задание.** Сравни возраст с числом 12 с помощью `>=`. Выведи полученный ответ.


```python
age = int(input("Сколько тебе лет? "))
print(age >= 12)
```

## 5. `if` и `else`: выбор по условию

`if` читается как «если». Если условие после `if` верно (`True`), Python выполняет строки с отступом под ним. Если условие неверно, можно выполнить другую ветку после `else` — «иначе».

Отступ показывает, какие команды относятся к ветке. В редакторе обычно нажимают Tab, но в этом блокноте отступ уже поставлен.

![Схема ветвления if/else](images/if-else.png)


```python
temperature = int(input("Сколько градусов на улице? "))

if temperature < 0:
    print("На улице мороз")
else:
    print("Температура не ниже нуля")
```

**Мини-эксперимент.** Запусти код дважды: введи сначала `-2`, затем `5`. Какую ветку выбрала программа?

**Задание.** Спроси число. Если оно равно 10, выведи «Нашли десять», иначе — «Это другое число».


```python
number = int(input("Введи число: "))

if number == 10:
    print("Нашли десять")
else:
    print("Это другое число")
```

## 6. Несколько вариантов: `elif`

`elif` означает «иначе, если». Он позволяет проверить ещё одно условие, когда предыдущее оказалось неверным. Python проверяет ветки сверху вниз и выполняет только первую подходящую. `else` в конце задаёт случай «ничего из перечисленного».


```python
score = int(input("Введи результат от 0 до 100: "))

if score >= 80:
    print("Отлично")
elif score >= 60:
    print("Хорошо")
else:
    print("Нужно ещё потренироваться")
```

Обрати внимание на порядок: сначала проверяется `score >= 80`, потом `score >= 60`. Если поставить проверку `>= 60` первой, результат 90 тоже подойдёт к ней, и до первой ветки выполнение уже не дойдёт.

**Мини-эксперимент.** Введи `90`, `70`, `40`. Затем поменяй местами условия и объясни результат.

**Задание.** Спроси температуру. Для значений ниже 0 напечатай «мороз», для значений от 0 до 20 включительно — «прохладно», иначе — «тепло».


```python
temperature = int(input("Температура: "))

if temperature < 0:
    print("Мороз")
elif temperature <= 20:
    print("Прохладно")
else:
    print("Тепло")
```

## 7. Соединяем условия: `and`, `or`, `not`

Иногда одного сравнения мало.

- `and` («и») даёт `True`, только если **оба** условия верны.
- `or` («или») даёт `True`, если верно **хотя бы одно** условие.
- `not` («не») меняет ответ на противоположный: `not True` — `False`, а `not False` — `True`.

Таблица показывает все сочетания двух условий A и B.

![Таблица истинности and or not](images/truth-table.png)


```python
print(True and True)
print(True and False)
print(False or True)
print(False or False)
print(not True)
```

**Мини-эксперимент.** Предскажи результат `False and True`, `False or True`, `not False`. Потом проверь.

**Задание.** Проверь, находится ли число в промежутке от 1 до 10 включительно: одновременно `number >= 1` и `number <= 10`.


```python
number = int(input("Введи число: "))
print(number >= 1 and number <= 10)
```

### Условия в одной программе

Логические операторы можно применять в `if`. Здесь билет выдаётся, если возраст не меньше 12 **и** есть приглашение, либо если возраст не меньше 16.


```python
age = int(input("Возраст: "))
has_invitation = input("Есть приглашение? Напиши да или нет: ")

if (age >= 12 and has_invitation == "да") or age >= 16:
    print("Можно войти")
else:
    print("Пока нельзя")
```

Скобки помогают увидеть, какие условия объединены вместе. Не забывай: ответ `input()` — текст, поэтому сравниваем его со строкой `"да"`.

**Мини-эксперимент.** Проверь сочетания: 10 лет с приглашением, 13 лет с приглашением, 16 лет без приглашения.


## 8. Короткая памятка

- `input()` получает строку; чтобы считать целое число, оберни ввод в `int(...)`.
- `=` сохраняет значение, `==` сравнивает.
- `if` проверяет условие, `elif` добавляет вариант, `else` задаёт запасной случай.
- Ветки выполняются сверху вниз; после первого совпадения остальные пропускаются.
- `and` требует, чтобы оба условия были верны; `or` — чтобы было верно хотя бы одно; `not` меняет ответ на противоположный.
- Тело каждой ветки отделяется отступом.
