## 1. Строка — текст в программе

**Строка** — это текст: слово, предложение или даже пустой текст. В Python строку записывают в кавычках. Подойдут одинарные `'...'` и двойные `"..."`; важно закрыть строку тем же видом кавычек, которым начали.

Кавычки помогают Python отличить текст программы от самого текста. Например, `"Привет"` — строка, а `Привет` без кавычек Python не поймёт как обычный текст.


```python
print("Привет!")
print('Сегодня урок Python.')
print("")  # пустая строка
```

## 2. Команда `print()` показывает результат

`print()` — команда для вывода информации. Она показывает на экране то, что написано внутри скобок. Текст помещают в кавычки. Скобки после `print` обязательны.

В одной команде `print()` можно передать несколько значений, разделив их запятыми. По умолчанию Python вставит между ними пробел.


```python
print("Мой город:", "Москва")
print("Мне нравится", "программировать")
```

**Мини-эксперимент.** Удали запятую между двумя частями текста и посмотри, что произойдёт. Верни запятую и поменяй слова местами.

## 3. Переменная хранит значение

**Переменная** — имя для значения, которое программа запоминает. Представь подписанную коробку: имя — это подпись, а внутри лежит значение.

Запись `city = "Казань"` означает: сохранить строку `"Казань"` под именем `city`. Знак `=` здесь означает «запомни справа под именем слева». Чтобы увидеть значение, передай имя переменной в `print()` без кавычек.


```python
city = "Казань"
favorite_color = "синий"
print(city)
print("Любимый цвет:", favorite_color)
```

### Как давать переменным понятные имена

Имя должно подсказывать, что хранится внутри: `name`, `school`, `favorite_game`. Если в имени несколько слов, разделяй их знаком подчёркивания: `favorite_color`.

В Python имена переменных обычно пишут маленькими латинскими буквами. Имя не должно начинаться с цифры, не должно содержать пробелы и не должно совпадать с командами языка вроде `print`. Например, `my_name` — хорошее имя; `2name` и `my name` — неправильные.


```python
student_name = "Аня"
lesson_topic = "строки"
print(student_name, "изучает", lesson_topic)
```

## 4. `input()` получает ответ

`input()` останавливает программу и ждёт, пока человек напечатает ответ и нажмёт Enter. Ответ можно сохранить в переменную. Важно: результат `input()` — **строка**, даже если человек ввёл цифры.

В скобках `input()` можно написать подсказку. Она помогает понять, что нужно ввести.


```python
name = input("Как тебя зовут? ")
print("Привет,", name)
```

## 5. Строки можно соединять и повторять

### Сложение строк

Оператор `+` соединяет строки подряд. Он не добавляет пробел сам: если он нужен, добавь его внутрь одной из строк.


```python
first = "Доброе "
second = "утро!"
print(first + second)
print("Доброе" + "утро!")  # слова соединятся без пробела
```

**Мини-эксперимент.** Добавь пробел в первую строку. Попробуй соединить три короткие строки.

### Умножение строки на число

Оператор `*` повторяет строку указанное целое число раз. Например, `"ха" * 3` даёт `"хахаха"`. Число ставится справа. Это повторяет именно текст, а не выполняет обычное умножение чисел.


```python
print("ха" * 3)
print("*" * 10)
```

## 6. f-строка вставляет значения в текст

Если нужно показать текст вместе со значениями переменных, можно использовать **f-строку**. Перед открывающей кавычкой ставится буква `f`, а имя переменной записывается в фигурных скобках `{}`.

Python подставит в это место значение переменной. Фигурные скобки в обычной строке без `f` остаются просто символами.


```python
name = "Миша"
hobby = "рисовать"
print(f"Меня зовут {name}. Я люблю {hobby}.")
print("Имя в скобках: {name}")
```

**Мини-эксперимент.** Измени значения `name` и `hobby`. Затем убери `f` перед кавычками в первой команде и сравни результат.

**Задание.** Спроси имя с помощью `input()`, сохрани его в `name`, а затем одной f-строкой напиши приветствие с этим именем.


```python
name = input("Как тебя зовут? ")
print(f"Рад знакомству, {name}!")
```

## 7. Настраиваем разделитель `sep` и окончание `end`

Когда в `print()` передано несколько значений через запятую, параметр `sep` задаёт текст между ними. По умолчанию это один пробел.

Параметр `end` задаёт, что добавить после вывода. По умолчанию Python заканчивает `print()` переходом на новую строку. Если задать `end=" "`, следующая команда `print()` продолжит вывод на той же строке.


```python
print("2026", "09", "26", sep="-")
print("Python", "урок", sep=" | ")

print("Привет", end=" ")
print("ещё раз!")
```

**Мини-эксперимент.** Измени `sep` на стрелку `" -> "`. Удали `end=" "` и посмотри, где окажется вторая часть.

**Задание.** Выведи три слова через звёздочку с помощью `sep`. Затем двумя командами `print()` выведи фразу в одной строке, используя `end`.


## 8. Специальные символы внутри строки

Некоторые символы записываются в строке через обратную косую черту `\` — это знак, похожий на наклонённую влево черту. Такая пара символов задаёт специальное действие:

- `\n` — перейти на новую строку;
- `\t` — вставить горизонтальную табуляцию, то есть широкий отступ;
- `\\` — показать одну обратную косую черту.

Обратная косая черта «сообщает» Python, что следующую букву нужно воспринимать особым образом. Например, `n` после `\` означает новую строку.


```python
print("Первая строка\nВторая строка")
print("Имя:\tАня")
print("Папка\\файл")
```

## 9. Итоговая практика

Собери короткую программу-знакомство. Она должна спросить имя и увлечение, а потом вывести аккуратную карточку на нескольких строках. Используй только то, что было в этом уроке: `input()`, переменные, f-строки, `print()` и `\n`.

**Подсказка:** сначала получи каждый ответ отдельно, затем составь одну f-строку с помощью `print()`.


```python
name = input("Как тебя зовут? ")
hobby = input("Что тебе нравится делать? ")
print(f"\n--- Знакомство ---\nИмя: {name}\nУвлечение: {hobby}\n-----------------")
```

### Проверь себя

- Строка — текст в кавычках.
- `print()` выводит данные, а `input()` получает ответ как строку.
- Переменная хранит значение под понятным именем.
- `+` соединяет строки, `*` повторяет строку.
- В f-строке перед кавычками стоит `f`, а значения переменных записывают в `{}`.
- `sep` меняет разделитель значений, `end` — окончание вывода.
- `\n`, `\t`, `\\` задают новую строку, отступ и обратную косую черту.

Если какая-то ячейка не запускается, проверь кавычки и скобки. Убедись, что строка начинается и заканчивается подходящими кавычками.
